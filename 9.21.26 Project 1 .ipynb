**Introduction**

This project explores New York City Regents exam performance from the 2014–15 through 2016–17 school years. The main question is: How does Brooklyn Technical High School’s mean Regents exam score compare with the mean Regents exam score from other NYC schools?

The mean test score will be used as the primary metric for comparison.

**Importance**

This analysis is important because it provides context for understanding Brooklyn Technical High School’s test performance relative to other schools across the city.

Python and pandas will be used to clean, analyze, and visualize the data.

**A.Import Libraries**


```python
# Import Libraries (Pandas)
import pandas as pd
import matplotlib.pyplot as plt
```

**B. Load the DataSet**


```python
# Load the Dataset
df = pd.read_csv("https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/2014-15_To_2016-17_School-_Level_NYC_Regents_Report_For_All_Variables.csv")
```


**D. Create a Subset**

```python
# Create Subset
# Show Names of Columns
df.columns
```

```python
# Create New Table by Pulling 3 Columns
df2 = df[['School Name', 'Regents Exam', 'Mean Score']]
```

**E. Data Cleaning**

```python
# Check for Missing Values
df2.isnull().sum()
```

```python
# Remove Missing Values
df2 = df2.dropna()
```

```python
# Confirm Missing Values Were Dropped
df2.isnull().sum()
```

**F.Filter**

```python
#Filter
#Filter Dataset for Brooklyn Tech High School
brooklyn_tech = df2[df2['School Name'] == 'Brooklyn Technical High School']
```

```python
#Data Exploration
#Top 5 Results of Subset Table
brooklyn_tech.head()
```

```python
#Data Exploration
## Descriptive Statistics: Summary of Table
brooklyn_tech.describe()
```

```python
#Data Exploration
# Check Data Types and Non-Null Values
brooklyn_tech.info()
```

**G. Subset: Descriptive Statistics**


```python
# Descriptive Statistics: Summary of Mean Score Before Numeric Conversion
brooklyn_tech['Mean Score'].describe()
```

```python
# Convert Mean Score From String to Numeric Value
df2['Mean Score'] = pd.to_numeric(df2['Mean Score'], errors='coerce')
```

```python
# Check for Missing Values After Numeric Conversion
df2.isnull().sum()
```

```python
# Remove Missing Values After Numeric Converion
df2 = df2.dropna()
```

```python
# Recreate Brooklyn Tech Subset Table After Data Cleaning
brooklyn_tech = df2[df2['School Name'] == 'Brooklyn Technical High School']
```

```python
# Descriptive Statistics: Summmary of Cleaned Brooklyn Tech Table
brooklyn_tech['Mean Score'].describe()
```

```python
# Calculate Brooklyn Tech (Average) Mean Score
brooklyn_tech['Mean Score'].mean()
```

```python
# Calculate Brooklyn Tech Median Mean Score
brooklyn_tech['Mean Score'].median()
```

**H. NYC Schools: Descriptive Statistics**

```python
# Descriptive Statistics: Summary of NYC Mean Score
df2['Mean Score'].describe()
```

```python
# Calculate NYC Schools Mean Score
df2['Mean Score'].mean()
```

```python
# Calculate NYC Schools Median Mean Score
df2['Mean Score'].median()
```

## **I. Compare the Results**

```python
# Calculate Brooklyn Tech and NYC Schools' Average Mean Score
brooklyn_mean = brooklyn_tech['Mean Score'].mean()
nyc_mean = df2['Mean Score'].mean()
```

```python
# Display Comparison
print("Brooklyn Technical High School Mean Score:", brooklyn_mean)
print("NYC Schools Mean Score:", nyc_mean)
```


**J. Visualize**

```python
# Visualize Brooklyn Technical High School Regent Mean Score Distribution
brooklyn_tech['Mean Score'].plot(kind='hist', color= 'hotpink', title='Brooklyn Technical High School: Regents Mean Scores',
    xlabel='Mean Score',
    ylabel='Frequency')
```

```python
# Visualize NYC Schools Mean Score Distribution
df2['Mean Score'].plot(kind='hist', color= 'red', title='NYC Schools: Regents Mean Scores', xlabel='Mean Score', ylabel='Frequency')
```

**K. Result**


```python
# Display Comparison
print("Brooklyn Tech Mean:", brooklyn_mean)
print("NYC Mean:", nyc_mean)
```
