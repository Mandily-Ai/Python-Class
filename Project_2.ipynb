```python
#Import Pandas
import pandas as pd
```

```python
#Read Dataset
df = pd.read_csv(
    'https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/mtcars.csv'
)
```

```python
#Understand the Data First
# Rows
df.head()
```

```python
# Understand the Data First
# Describe
df.describe()
#Result: 32 MPG observations, Mean (average)= 20.09
```

```python
#Understand the Data First
#Columns
df.info()
#Result: 32 entries, 12 Columns, Column Names, Non-Null Counts, DataType
```

```python
#Clean the Data Set
#Rename the unnamed column
df.rename(columns={'Unnamed: 0': 'car'}, inplace=True)
```

```python
#Create new data frame
#Result: Created new dataframe named df2 (from df)
#Result: Contains only 4 columns (mpg, hp, wt, cycle)
#Reads: df2 equals df, selecting only mpg, hp, wt, cycl
df2 = df[['mpg', 'hp', 'wt', 'cyl']]
```

```python
# Rename the four columns
# Reads as: Rename mpg= miles per gallon, hp = horsepower, wt= weight, cyl = cylinders
df2 = df2.rename(columns={
    'mpg': 'miles_per_gallon',
    'hp': 'horsepower',
    'wt': 'weight',
    'cyl': 'cylinders'
})
```

```python
#Summary Statistics for those columns
# Use Describe again
# Reads: Summary of only those 4 columns
# Result: Gives averages and counts of 4 columnns (variables)
df2.describe()
```

```python
#Read
df2.head()
```

```python
#Filter your Data plus Find Summary Statistics for Only One Type of Vehicle
#Create a new subset (create a new dataframe called four_cycl)
#Contains one type of vehicle =
#(Contains only the rows from df2 where the cylinders column equals 4)

four_cyl = df2[df2['cylinders'] == 4]
```
