```python
#Create 2-3 Visualizations from the Dataset
#Question: What Borough has the most complaints by Borough
#Result: Brooklyn has the most 311 complaints
df['Borough'].value_counts().plot(kind='bar', color='pink', title='311 Complaints by Borough')
```

```python
#Create a Plot of Counts by Borough by Noise Complaints
#Question: Which borough has the most noise complaints?
#Result: The line plot shows that Manhattan has the
# highest number of noise complaints. (unique values not filtered)


df[df['Complaint Type'] == 'Noise']['Borough'].value_counts().plot(kind='line',
    color='limegreen', title='Noise Complaints by Borough')

```
